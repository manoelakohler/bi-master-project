```
import math
import Hive
import Utilities
```

image.png

```
# funcao de avaliação
def objective_function(vector):
  acum = 0
  for var in vector:
    acum += var**2 - 10*math.cos(2*math.pi*var) + 10
  return acum
```

```
# cria modelo
ndim = 5
model = Hive.BeeHive(lower     = [-5.12]*ndim,
                     upper     = [5.12]*ndim,
                     fun       = objective_function,
                     numb_bees =  100,
                     max_itrs  =  100)

# runs model
cost = model.run()

# plots convergence
Utilities.ConvergencePlot(cost)
```

```
# prints out best solution
print("Fitness Value ABC: {0}".format(model.best))
```
