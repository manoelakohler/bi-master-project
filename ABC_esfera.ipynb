```python
import Hive
import Utilities
```

Função Objetivo

```python
def objetctive_function(bee):
  acum = 0
  for var in bee:
    acum += var**2
  return acum
```

```python
objetctive_function([0,0])
```

```python
# Criar o modelo
ndim = int(5)
model = Hive.BeeHive(lower      = [-600]*ndim,
                      upper     = [600]*ndim,
                      fun       = objetctive_function,
                      numb_bees =  30,
                      max_itrs  =  100)
```

```python
# Rodar o modelo
results = model.run()
```

```python
results.keys()
```

```python
# Plot do processo evolutivo
Utilities.ConvergencePlot(results)
```

```python
# imprime a melhor solução
print("Fitness Value ABC: {0}".format(model.best))
```
